# 1. Timing Code

**Rubric:** *Student calls the time functions before the start of main code and after the main logic has been finished.*

`check_images.py` needs to report how long the whole classification run took. `time.time()` returns the current time in seconds since the epoch, so calling it once right before the main logic starts and once right after it finishes -- and subtracting the two -- gives the total elapsed runtime.

This notebook shows the pattern in isolation, then confirms it is wired up correctly inside `check_images.py`.

````python
import sys, os
sys.path.insert(0, os.path.abspath('..'))  # so we can import the project's .py files
````

## The pattern

````python
from time import time, sleep

# --- Start timer, BEFORE the main logic ---
start_time = time()

# --- "main logic" (stands in for get_pet_labels/classify_images/etc.) ---
sleep(1.2)

# --- Stop timer, AFTER the main logic has finished ---
end_time = time()

tot_time = end_time - start_time
print(f"Elapsed seconds: {tot_time:.2f}")
````

## Converting to hh:mm:ss

A raw seconds count isn't very readable for a run that might take several minutes, so `check_images.py` converts it to `hours:minutes:seconds` before printing it.

````python
hours = int(tot_time / 3600)
minutes = int((tot_time % 3600) / 60)
seconds = int(tot_time % 60)
print(f"** Total Elapsed Runtime: {hours:02d}:{minutes:02d}:{seconds:02d}")
````
