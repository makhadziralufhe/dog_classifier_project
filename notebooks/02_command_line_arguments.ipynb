# 2. Command Line Arguments

**Rubric:**
- Q1: enable `--dir`, default `'pet_images/'`
- Q2: enable `--arch`, default `'vgg'`
- Q3: enable `--dogfile`, default `'dognames.txt'`

`get_input_args()` in `get_input_args.py` uses `argparse` to define these three options so `check_images.py` can be run as:

```bash
python3 check_images.py --dir pet_images/ --arch vgg --dogfile dognames.txt
```

or with no arguments at all, in which case every default above applies.

````python
import inspect
from get_input_args import get_input_args
print(inspect.getsource(get_input_args))
````

## Checking the defaults

`sys.argv` is what `argparse` reads from; setting it to just the program name (no flags) lets us check that every default matches the rubric exactly, without needing an actual terminal.

````python
import sys

sys.argv = ['check_images.py']  # simulate running with no flags
in_arg = get_input_args()

print('in_arg.dir     =', repr(in_arg.dir))
print('in_arg.arch    =', repr(in_arg.arch))
print('in_arg.dogfile =', repr(in_arg.dogfile))

assert in_arg.dir == 'pet_images/'
assert in_arg.arch == 'vgg'
assert in_arg.dogfile == 'dognames.txt'
print('\nAll three defaults match the rubric.')
````

## Overriding each one from the command line

````python
sys.argv = ['check_images.py', '--dir', 'pet_images/', '--arch', 'resnet', '--dogfile', 'dognames.txt']
in_arg = get_input_args()
print(in_arg)
````

`--arch` is restricted to the three real architectures this project supports, so a typo is caught immediately instead of failing deep inside `classify_images()`:

````python
sys.argv = ['check_images.py', '--arch', 'not-a-real-model']
try:
    get_input_args()
except SystemExit:
    print('argparse rejected the bad --arch value, as expected.')
````
