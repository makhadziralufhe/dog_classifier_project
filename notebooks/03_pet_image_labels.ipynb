`get_pet_labels()` treats each filename in `pet_images/` as ground truth: it strips the extension and the trailing digits, turns underscores into spaces, and lowercases everything to recover the pet's true label.

## Running it against `pet_images/`

````python
results_dic = get_pet_labels('../pet_images/')
print('Number of key-value pairs:', len(results_dic))

import itertools
for filename, label in itertools.islice(results_dic.items(), 8):
    print(f'{filename:28s} -> {label}')
````

````python
assert len(results_dic) == 40, 'expected exactly 40 pet images'
for filename, value in results_dic.items():
    assert isinstance(value, list) and len(value) == 1, filename
print('All 40 entries are single-element lists, matching the required format.')
````
